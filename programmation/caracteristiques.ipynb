```python
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Analyse mathématique

### 2.1. Méthode des caractéristiques

```python
# Avec conditions au bord
def caracteristique(X, t):
    return X + t
    
t = np.linspace(0, 2, 1000)

plt.figure(figsize=(10, 7))

for X in [-1/2, -1/4, 0, 1/4, 1/2, 3/4, 1, 5/4, 6/4]:
    plt.plot(caracteristique(X, t), t, color='blue')

plt.xlabel("$a$")
plt.xlim(0, 4)
plt.ylabel("$t$")
plt.title("Courbes caractéristiques dans le demi-plan $(a,t)$")
plt.grid(True)
plt.show()
```

```python
# Sans conditions au bord
def caracteristique(X, t):
    return X + t
    
t = np.linspace(0, 2, 1000)

plt.figure(figsize=(10, 7))

for X in [-1/2, -1/4, 0, 1/4, 1/2, 3/4, 1, 5/4, 6/4]:
    plt.plot(caracteristique(X, t), t, color='blue')

plt.axvline(x=0, color='black', linestyle='--')

plt.xlabel("$a$")
plt.ylabel("$t$")
plt.title("Courbes caractéristiques dans le demi-plan $(a,t)$")
plt.grid(True)
plt.show()
```
